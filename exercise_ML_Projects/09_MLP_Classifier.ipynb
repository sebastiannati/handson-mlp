# Chapter 10: Train a MLP classifier on CoverType dataset

## Get the data

```python
from sklearn.datasets import fetch_covtype
from sklearn.model_selection import train_test_split

data = fetch_covtype(as_frame=True)
print(data.DESCR)

X_train, X_test, y_train, y_test = train_test_split(data.data, data.target,random_state=42,test_size=0.20)
print(X_train.shape)
print(X_test.shape)
```

## Analyse and prepare data

```python
X_train.head()
```

We can see that values in dataset have very different scales, so for our mlp we'll need to scale them.

```python
from sklearn.pipeline import Pipeline
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler


mlp_clf = MLPClassifier(hidden_layer_sizes=[300,100,50],random_state=42,early_stopping=True,verbose=1)

pipeline = Pipeline(
    steps=[("Scaler", StandardScaler()), ("MLP", mlp_clf)]
)

pipeline.fit(X_train, y_train)
```

```python
pipeline.score(X_test,y_test)
```
